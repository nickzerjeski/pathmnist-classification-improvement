# Experiment 2: Halder-Style ViT Fine-Tuning

This notebook loads saved artifacts from `results/experiment2` and `models/experiment2`. It does not retrain the model.

Reference target: ViT-Base-Patch16-224 PathMNIST ACC `0.9462`, cancer precision `0.97`, cancer recall `0.95`, cancer F1 `0.96`.

```python
import json
from pathlib import Path
import pandas as pd

RESULT_DIR = Path('../results/experiment2')
metrics_path = RESULT_DIR / 'metrics.json'
if not metrics_path.exists():
    raise FileNotFoundError('Run experiment2 from data/experiment-design.md first.')
metrics = json.loads(metrics_path.read_text())
test = metrics['test']
pd.Series({k: v for k, v in test.items() if isinstance(v, (int, float))})
```

```python
passed = (
    test['acc'] >= 0.9462
    and test['cancer_precision'] >= 0.97
    and test['cancer_recall'] >= 0.95
    and test['cancer_f1'] >= 0.96
)
print('Passed Halder ViT gate:', passed)
```
